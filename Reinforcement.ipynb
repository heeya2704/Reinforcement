## Question 1: Differences Between Supervised, Unsupervised, and Reinforcement Learning

### Key Differences Summary:
1. **Data & Supervision:** Supervised learning uses labeled input-output pairs (X, y); Unsupervised learning uses unlabeled data X to find hidden structures; Reinforcement Learning (RL) has no static dataset or explicit labels and learns via dynamic interaction with an environment using state, action, and reward signals.
2. **Feedback Mechanism:** Supervised learning gets immediate per-sample ground truth feedback; Unsupervised learning receives no external feedback; RL receives delayed or sequential feedback through reward (+R) and penalty (-R) signals over an episode.
3. **Objective:** Supervised aims to minimize prediction error; Unsupervised aims to find data clusters or latent patterns; RL aims to discover an optimal policy that maximizes cumulative long-term reward.

### Comparison Table with Real-World Examples:

| Feature / Aspect | Supervised Learning | Unsupervised Learning | Reinforcement Learning |
|---|---|---|---|
| **Primary Mechanism** | Learns from labeled datasets (X, y) with explicit correct answers. | Analyzes unlabeled data X to discover underlying patterns or groupings. | Learns optimal decision-making through trial-and-error interaction with an environment. |
| **Feedback Type** | Direct, immediate error feedback against true target labels. | No external feedback (uses intrinsic measures like variance/distance). | Delayed or sequential reward (+) and penalty (-) feedback over time. |
| **Instagram Example (Feed Ranking)** | Predicting if a user will like a post (Features: past likes, post category -> Output: 1 or 0). | Clustering Instagram users into interest groups (e.g., Fashion, Tech, Travel) using K-Means. | RL agent dynamically sequencing the feed to maximize user app retention time (+reward for long view time/likes, -penalty for fast scrolls). |
| **Spotify Example (Music Recs)** | Predicting a user's rating (1–5 stars) for a new track based on historical listening preferences. | Grouping tracks into acoustic clusters/moods (e.g., 'Chill Lofi', 'Workout EDM') using audio features without genre tags. | RL agent generating a dynamic song queue to maximize listening session duration (+reward for full play, -penalty for skip). |
| **Chess AI Example** | Predicting move choices trained on Grandmaster historical game logs (Input: Board state -> Output: Expert Move). | Clustering common opening board positions or strategies into tactic categories without outcome labels. | RL Agent (like AlphaZero) playing millions of self-play games to learn winning strategy (+1 for Win, -1 for Loss, 0 for Draw). |

## Question 2: Step-by-Step RL Application — Zomato Restaurant Recommendation Engine

### Application Overview:
A personalized restaurant recommendation engine on Zomato that dynamically curates and ranks restaurant choices for users opening the app at lunchtime or dinner.

```
                     +--------------------+ 
                     | Zomato RL Agent    |
                     +---------+----------+
                               | Action: Display Top 5 Ranked Restaurants
                               v
                     +--------------------+
                     | App Environment &  |
                     | User Behavior      |
                     +---------+----------+
                               | State: User context, time, location
                               | Reward: Click (+1), Order (+10), App Exit (-5)
                               v
                     +--------------------+
                     | RL Agent Updates   |
                     | Policy (Q-Learning)|
                     +--------------------+
```

### RL Component Breakdown:
1. **Agent:**
   * The **Zomato Recommendation Engine Agent**, which is responsible for selecting, ranking, and displaying restaurant options and banners to maximize user engagement and completed orders.
2. **Environment:**
   * The Zomato App Interface, User Context, and Real-time User Behavior (time of day, current location, local weather, dietary filters, price tier).
3. **State (S_t):**
   * User profile vector (past order history, favorite cuisines, average order value).
   * Contextual features: Time = 1:30 PM (Lunch), Location = Connaught Place, Weather = Rainy.
   * Current session state (search query, applied filters).
4. **Action (A_t):**
   * Selecting a list of top 5 recommended restaurants to show on the user's home screen.
   * Pairing specific promotional deals/discounts with those restaurants (e.g., '60% OFF up to 120').
5. **Reward Signal (R_t):**
   * **+1 point:** User clicks on a recommended restaurant card.
   * **+3 points:** User adds item to cart.
   * **+10 points:** User completes an order successfully.
   * **+5 points:** High rating given by user post-delivery.
   * **-1 point:** User scrolls past without clicking.
   * **-5 points:** User closes/exits the app without placing any order.
6. **Learning Process:**
   * The agent balances **Exploration** (trying newly onboarded restaurants or new cuisines) and **Exploitation** (recommending the user's historical favorites) to continuously increase total completed orders and user retention over time.

## Question 3: Simple RL Agent for IPL Fantasy Cricket Game App

### Scenario Overview:
Designing an RL agent that acts as an automated Fantasy Manager, assembling the optimal 11-player squad for upcoming IPL matches under official fantasy league constraints.

### Component Specifications:

#### 1. Agent:
* **IPL Fantasy Squad Manager Agent** — tasked with selecting squad lineups and appointing Captain & Vice-Captain to maximize total fantasy match points.

#### 2. Environment:
* The IPL Tournament & Player Statistics Environment. It enforces constraints:
  * 100 Credit Point Budget Limit.
  * Squad composition: 11 players total (e.g., 3-5 Batsmen, 1-2 Wicketkeepers, 1-3 All-Rounders, 3-5 Bowlers).
  * Max 4 Overseas players.
  * Pitch conditions (e.g., spin-friendly, pace/bounce pitch), venue history, weather, and recent player form.

#### 3. State (S_t):
* Remaining budget credits.
* Current selected squad composition (count of batsmen, bowlers, overseas slots used).
* Opponent team characteristics, pitch report, and player availability (Playing XI status).

#### 4. Possible Actions (A_t):
* `ADD_PLAYER(Player_ID)`: Add an available player to the squad.
* `REMOVE_PLAYER(Player_ID)`: Drop a player to free up credits/slots.
* `SET_CAPTAIN(Player_ID)`: Assign Captain role (2x point multiplier).
* `SET_VICE_CAPTAIN(Player_ID)`: Assign Vice-Captain role (1.5x point multiplier).

#### 5. Reward Assignment Scheme (R_t):
* **Match Performance Reward:** Total fantasy points earned by the selected 11 players in real match simulation:
  * Batting: +1 pt per run, +1 per boundary, +2 per six, +8 for half-century bonus.
  * Bowling: +25 pts per wicket, +8 for LBW/Bowled bonus, +12 for 4-wicket haul.
  * Fielding: +8 pts per catch/stumping.
  * Captain & Vice-Captain Multipliers applied to individual points.
* **Rule Violation Penalties:**
  * **-100 penalty:** Exceeding 100 credit budget limit or invalid team structure (e.g., selecting 0 wicketkeepers).
  * **-50 penalty:** Selecting a player who is benched / not in the official match Playing XI.

## Question 4: Enhancing an App Feature with RL — YouTube Autoplay / Recommendation Feature

### Feature Chosen:
**YouTube Autoplay & Next-Video Recommendation System**

### Problem with Existing Supervised Systems:
Traditional supervised learning models optimize for immediate Click-Through Rate (CTR). This often causes recommending clickbait videos or repetitive single-topic traps. Users click out of curiosity, get disappointed, and quickly churn from the app.

### How Reinforcement Learning Makes it Smarter:
RL treats video recommendation as a multi-step sequential decision problem. Instead of asking *'Which video has the highest probability of a click next?'*, the RL agent asks *'Which sequence of videos will maximize total user satisfaction and watch time over a 60-minute session?'*

* **State (S_t):** Currently playing video metadata, user history, time spent on the last 5 videos, time of day, device type, network connection.
* **Action (A_t):** Selecting the next video to autoplay (or top candidates for sidebar queue).

### Reward Signal (R_t) Design:
* **+10 High Positive Reward:** User watches >80% of the recommended video or watches until completion.
* **+5 Engagement Reward:** User likes, comments, shares, or subscribes to the channel during/after the video.
* **-2 Minor Penalty:** User skips video within the first 30 seconds.
* **-10 Major Penalty:** User turns off Autoplay feature or immediately exits/closes the YouTube app.
* **+20 Long-term Session Reward:** User stays engaged and continues watching across a multi-video session exceeding 30 minutes without churning.
